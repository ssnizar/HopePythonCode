```python
def oddEven():
    num1=input("Enter the number:")
    if not (num1.isdigit()):
        #print("No Number")
        Messge="No Number"
        return Messge
    
    num=int(num1)
    if ((num%2)==0):
        Messge="Even Number"
        #print("Even Number")
    else:
        Messge="Odd Number"
        #print("Odd Number")
    return Messge

result=oddEven()
if (result=="Even Number"):
   print("The given number is Even Number");
elif (result=="Odd Number"):
   print("The given number is Odd Number");
else:
    print("The are not provided Number ");
   
```
