```python
#3.Write a Python program to print the sum of numbers from 1 to 10. 
#Expected Output:55

SumofNumber=10
SumNumber=0
for loopNo3 in range(1,SumofNumber+1):
    SumNumber=SumNumber+loopNo3

print(SumNumber)
```

```python
#4.Write a Python function to find the factorial of 5. Expected Output:
#120

def factorial(number):
    if (number==0 or number==1):
        return 1
    else:
        return number * factorial(number-1)

print(factorial(5))
```

```python
#5.Write a Python program to check if a number is prime. Expected Output (for input 7):
#7 is a prime number

def isPrime(number):
    for pn in range(2, (int(number**0.5))+1):
        if (number % pn ==0):
            return False
    return True

myVerify_No=7
if isPrime(int(myVerify_No)):
    print(f"{myVerify_No} is Prime number")
else:
    print("it is not prime number")
```
