```python
transAmount=input("enter the transaction amount")
if (int(transAmount)==0):
    print("Transacton amount is Zero")
elif (int(transAmount)>0):
    print("Transaction amount is Positive")
else:
    print("Transaction amount is Negative")
```

```python

passCode=input("enter the passCode")
sum1=0
if (passCode.isnumeric()):
    print("Verified, it is  numberic number")
    for pcNo in passCode:
        sum1=sum1+int(pcNo)
    print(sum1)
else:
    print("Verified, it is not numberic number")
```
