```python
transactionAmount=input("Enter the Transaction Amount")
if (transactionAmount.lstrip('-').isdecimal()):
    #print("Digit")
    if(int(transactionAmount)>0):
        print(f"{transactionAmount} is Positive")
    elif(int(transactionAmount)<0):
        print(f"{transactionAmount} is Negative")                
    else:
        print("Transaction Amount is Zero")                
else:
    print("Not Digit")
```

```python
#sum of the digit for passcode
passCode=input("enter the passCode")
sum1=0
if (passCode.isdecimal()):
    print("Verified, it is  numeric number")
    for pcNo in passCode:
        sum1=sum1+int(pcNo)
    print("sumn of the numberic number for passcode",sum1)
else:
    print("Not Valid, Only allows Numbers")
    
```

```python
#logic check given number is prime

def isPrimeNo(number):
    #print(number/2)
    #print(number ** 0.5)
    for verN in range(2, (int(number**0.5))+1):
        if (number % verN == 0):
            return False 
    return True

primeNoVar=input("enter the PrimeNo.")
if (primeNoVar.lstrip('-').isdecimal() and int(primeNoVar)>1 ):
    if (isPrimeNo(int(primeNoVar))):
        print(f"{primeNoVar} is Prime Number")                
    else:
        print(f"{primeNoVar} is Not Prime Number")                
else:
    print("Not Digit")
```

```python
#factorial and primeNoVar=input("enter the PrimeNo.")

def factorialNumber(number):
    if (number==0 or number==1):
        return 1    
    else:
        return number * factorialNumber(number-1)
    
FactNoVar=input("enter the PrimeNo.")
if (FactNoVar.lstrip('-').isdecimal() and int(FactNoVar)>1 ):
    myfactvalue=factorialNumber(int(FactNoVar))
    print(f"{FactNoVar} for factorial is {myfactvalue}")
else:
    print("Not Digit")

```

```python
#armstrong number 153,370,,371,1634
def findArmstronint(strNumber):
    powerVal=int(len(strNumber))
    ArmNoSum=0
    for ArmNo in strNumber:
        myvalue=(int(ArmNo))**powerVal
        ArmNoSum=ArmNoSum+myvalue
    return ArmNoSum

ArmsstrongNoVar=input("enter the ArmsstrongNo.")
if (ArmsstrongNoVar.lstrip('-').isdecimal() and int(ArmsstrongNoVar)>1 ):
    myArmvalue=findArmstronint(ArmsstrongNoVar)
    if (myArmvalue==int(ArmsstrongNoVar)):
        print("Given Value is Armstrong Value")
    else:
        print("Given Value is NOT Armstrong Value")
else:
    print("Not Digit")
```

```python
#most largest words
SentValue=input("pleasprovide the sentence to find big word")
WordList=SentValue.split()
#print(WordList)
Hword=""
HwordLen=0
for word1 in WordList:
    tWordLen=len(word1)
    if  tWordLen > HwordLen:
        HwordLen=tWordLen
        Hword=word1

print(f"Given Sentence Longest word is :  {Hword}")
```

```python
#anagrams
#Listen - Silent / Earth  - Heart / Below  Elbow  /Tom Marvolo Riddle -  I am Lord Voldemort
def isanagram(str1, str2):
    # Optional: convert to lowercase and remove spaces for accuracy
    str1 = str1.replace(" ", "").lower()
    str2 = str2.replace(" ", "").lower()
    
    return sorted(str1) == sorted(str2)

s1=input("enter the 1st sentence for anagaram")
s2=input("enter the 2nd sentence for anagaram")

if isanagram(s1,s2):
    print("Both sentence of anagarm sentences")
else:
    print("It is not anagarm sentences")
```
